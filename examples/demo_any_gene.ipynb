# tf-explorer Demo

This notebook demonstrates how to use `tf-explorer` to analyze transcription factor binding sites for **ANY GENE**.

We will analyze **PWWP2A** as a primary example, and then show how to easily switch to other genes like **TP53**.

```python
import os
import pandas as pd
from tf_explorer import analysis, motifs

# Setup output directory
OUT_DIR = "demo_results"
os.makedirs(OUT_DIR, exist_ok=True)
```

## 1. Analyze PWWP2A

We will search for binding sites of **E2F1** and **YY1**.

```python
gene = "PWWP2A"
tfs = ["E2F1", "YY1"]
# Using common IDs for E2F1 and YY1
jaspar_ids = [motifs.COMMON_TFS["E2F1"], motifs.COMMON_TFS["YY1"]]

print(f"Analyzing {gene}...")
analysis.analyze_gene(
    gene_name=gene,
    tf_list=tfs,
    jaspar_ids=jaspar_ids,
    output_dir=os.path.join(OUT_DIR, gene),
    bed_output=True,
    plot_track=True
)
```

### View Results
Let's look at the summary.

```python
summary_file = os.path.join(OUT_DIR, gene, f"{gene}_combined_summary.csv")
if os.path.exists(summary_file):
    print(pd.read_csv(summary_file))
else:
    print("Analysis failed or no output generated.")
```

## 2. Analyze TP53 (Any Gene!)

Now let's try a completely different gene, **TP53**, and check for **CTCF** binding.

```python
gene = "TP53"
tfs = ["CTCF"]
jaspar_ids = [motifs.COMMON_TFS["CTCF"]]

print(f"Analyzing {gene}...")
analysis.analyze_gene(
    gene_name=gene,
    tf_list=tfs,
    jaspar_ids=jaspar_ids,
    output_dir=os.path.join(OUT_DIR, gene),
    bed_output=True,
    plot_track=True
)
```

```python
summary_file = os.path.join(OUT_DIR, gene, f"{gene}_combined_summary.csv")
if os.path.exists(summary_file):
    print(pd.read_csv(summary_file))
```
